The data points are generated by the number of points required, If we need 10 points
We would first divide n by the number range we are requiring to get points, here it is in 8, So we divide by 8/n and use that difference and add from 0 + diff = new_value , new_value + diff = new_value_2, ......, new_value_n + diff = 8 , So we each new value is a data point.
