```python
import matplotlib.pyplot as plt
import random
import math
## Create the Area
fig, ax = plt.subplots()
ax.set_xlim(-10,10)
ax.set_ylim(-10,10)
ax.grid(True)
ax.set_xlabel("X-axis")
ax.set_ylabel("Y-axis")

## Make a Square
x_dp_squ = [8,-8,-8,8,8]
y_dp_squ = [8,8,-8,-8,8]
plt.plot(x_dp_squ,y_dp_squ,color='blue',linewidth = 2)

## Make a Circle
circle = plt.Circle((0,0),4,color='red',fill=False,linewidth=2)
plt.gca().add_patch(circle)

plt.show

## Make data points
x_data_pts = []
y_data_pts = []
n = int(input("Number of points: "))
for i in range(n):
    x = random.randint(-8, 8)
    y = random.randint(-8, 8)
    
    x_data_pts.append(x)
    y_data_pts.append(y)

print(x_data_pts)
print(y_data_pts)
    
plt.plot(x_data_pts,y_data_pts,'o')


## Calculate the distances from center (0,0) and append those points that lie within the perimeter of the Circle 
Circle = []
for i in range(len(x_data_pts)):
    r = math.sqrt((x_data_pts[i])**2 + (y_data_pts[i])**2)
    print(round(r,2))
    if round(r,2)<=4:
        Circle.append(r)
    else:
        pass
print(Circle,len(Circle))
print(16*(len(Circle)/n))
```
